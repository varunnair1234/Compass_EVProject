```python
import torch
import torch.nn as nn
import os

# 1. WE MUST REDEFINE THE ARCHITECTURE (The Skeleton)
# PyTorch needs to know the shape of the brain before it can load the memories
class PreferenceCalibrationNet(nn.Module):
    def __init__(self):
        super(PreferenceCalibrationNet, self).__init__()
        self.fc1 = nn.Linear(2, 16)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(16, 3)
        self.softmax = nn.Softmax(dim=1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        x = self.softmax(x)
        return x

# 2. INITIALIZE A BLANK MODEL AND LOAD THE MEMORIES
model_path = '../saved_models/preference_calibrator.pth'

# Create a fresh, dumb model
production_model = PreferenceCalibrationNet()

# Inject the trained intelligence
if os.path.exists(model_path):
    production_model.load_state_dict(torch.load(model_path))
    production_model.eval() # Lock the model into prediction mode
    print("✅ Model successfully loaded from disk and ready for production!")
else:
    print("❌ Error: Could not find the .pth file. Did you run the save block in notebook 03?")

# 3. RUN A PRODUCTION INFERENCE TEST
# Let's test a middle-class user with an average commute
income_k = 75.0
commute_miles = 30.0

with torch.no_grad():
    user_tensor = torch.tensor([[income_k, commute_miles]], dtype=torch.float32)
    predicted_weights = production_model(user_tensor).numpy()[0]
    
    print("\n🔮 Production Prediction Test:")
    print(f"User Income: ${income_k}k | Commute: {commute_miles}m")
    print(f"  ↳ Price Weight:   {predicted_weights[0]*100:.1f}%")
    print(f"  ↳ Range Weight:   {predicted_weights[1]*100:.1f}%")
    print(f"  ↳ Battery Weight: {predicted_weights[2]*100:.1f}%")
```
