```python
import pandas as pd
import numpy as np
import os

# 1. LOAD THE RAW DATA
file_name = 'compiled_vehicle_models_data.csv'
# We use our 'nishi' search logic to find it anywhere on your Mac
search_folders = ['.', '..', '/Users/nishi/Downloads', '/Users/nishi/Desktop']

found_path = None
for folder in search_folders:
    p = os.path.join(folder, file_name)
    if os.path.exists(p):
        found_path = p
        break

if found_path:
    df_raw = pd.read_csv(found_path)
    print(f"✅ Raw Data Loaded from: {found_path}")
    print(f"📊 Shape: {df_raw.shape}")
    print("\n--- First 5 rows of Raw Data ---")
    print(df_raw.head())
else:
    print("❌ Could not find the CSV file.")
```

```python
# 2. DATA CLEANING & FEATURE ENGINEERING
if found_path:
    df_clean = df_raw.copy()
    
    # Standardize column names to lowercase for consistency
    df_clean.columns = [c.lower() for c in df_clean.columns]
    
    # Feature Engineering: Convert KM to Miles if necessary
    if 'range_km' in df_clean.columns and 'range_miles' not in df_clean.columns:
        df_clean['range_miles'] = df_clean['range_km'] * 0.621371
        print("📏 Converted Range (KM) to Range (Miles)")

    # Handle the scrambled Price column
    # Rename to a standard 'Price_USD' for the rest of the project
    if 'price_usd' in df_clean.columns:
        df_clean = df_clean.rename(columns={'price_usd': 'Price_USD'})
        
    # Drop any rows that are missing critical info (Price or Range)
    initial_count = len(df_clean)
    df_clean = df_clean.dropna(subset=['Price_USD', 'range_miles'])
    
    print(f"🧹 Dropped {initial_count - len(df_clean)} rows with missing data.")
    print(f"✨ Final Cleaned Dataset ready with {len(df_clean)} vehicles.")
    
    # Save the cleaned version for the other notebooks to use
    os.makedirs('../data', exist_ok=True)
    df_clean.to_csv('../data/cleaned_ev_data.csv', index=False)
    print("💾 Cleaned data saved to: ../data/cleaned_ev_data.csv")
```
