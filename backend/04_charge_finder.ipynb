```python
import requests

def get_nearest_fast_chargers(lat=36.9741, lon=-122.0308, radius_miles=15, limit=5):
    """
    Fetches the nearest EV DC Fast Chargers using the NREL API via GPS coordinates.
    """
    url = "https://developer.nrel.gov/api/alt-fuel-stations/v1/nearest.json"
    
    # Updated parameters using 'latitude' and 'longitude' instead of 'location'
    params = {
        "api_key": "DEMO_KEY",
        "latitude": lat,
        "longitude": lon,
        "radius": radius_miles,
        "fuel_type": "ELEC",
        "ev_charging_level": "dc_fast",
        "limit": limit
    }

    print(f"🔍 Searching for DC Fast Chargers near Lat: {lat}, Lon: {lon} (Santa Cruz area)...\n")
    
    try:
        response = requests.get(url, params=params)
        
        # Print exact error if it fails again
        if response.status_code != 200:
            print(f"❌ API Error {response.status_code}: {response.text}")
            return

        data = response.json()
        stations = data.get('fuel_stations', [])
        
        if not stations:
            print("No fast charging stations found in this area.")
            return

        for i, station in enumerate(stations, 1):
            name = station.get('station_name')
            address = station.get('street_address')
            city = station.get('city')
            distance = station.get('distance')
            network = station.get('ev_network')
            connectors = station.get('ev_connector_types')
            
            print(f"#{i}: {name} ({network})")
            print(f"   📍 Distance: {distance:.2f} miles")
            print(f"   🗺️ Address: {address}, {city}")
            print(f"   🔌 Plugs: {', '.join(connectors) if connectors else 'Unknown'}")
            print("-" * 40)
            
    except Exception as e:
        print(f"Request Failed completely: {e}")

# Run the function
get_nearest_fast_chargers()
```
